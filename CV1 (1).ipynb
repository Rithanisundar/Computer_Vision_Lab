```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Read the image
I = mpimg.imread("C:\\Users\\student\\Downloads\\pexels-suissounet-2101187 (1).jpg")

# Extract the red, green, and blue components
R = I[:, :, 0]
G = I[:, :, 1]
B = I[:, :, 2]

# Create subplots
plt.figure(figsize=(10, 10))

# Display the red component
plt.subplot(2, 2, 2)
plt.imshow(R, cmap='Reds')
plt.xlabel('RED component is extracted')
plt.grid(True)

# Display the green component
plt.subplot(2, 2, 3)
plt.imshow(G, cmap='Greens')
plt.xlabel('GREEN component is extracted')
plt.grid(True)

# Display the blue component
plt.subplot(2, 2, 4)
plt.imshow(B, cmap='Blues')
plt.xlabel('BLUE component is extracted')
plt.grid(True)

# Show the plots
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

def colouredToGray(img):
    # Extract Red, Green, and Blue components
    R = img[:, :, 0]
    G = img[:, :, 1]
    B = img[:, :, 2]

    # Get the number of rows (M) and columns (N) of the RGB image matrix
    M, N, _ = img.shape

    # Create a new 2D matrix 'gray_img' of size M*N with all elements as zero
    gray_img = np.zeros((M, N), dtype=np.uint8)

    # Calculate grayscale values by forming a weighted sum of the R, G, and B components for each pixel
    for i in range(M):
        for j in range(N):
            gray_img[i, j] = int((R[i, j] * 0.2989 + G[i, j] * 0.5870 + B[i, j] * 0.114) * 255)

    return gray_img

# Read the image
img = mpimg.imread("C:\\Users\\student\\Downloads\\lenna-png-1.png")


# Convert the image to grayscale
gray_img = colouredToGray(img)

# Display the grayscale image
plt.imshow(gray_img, cmap='gray')
plt.xlabel('Grayscale Image')
plt.show()
```
