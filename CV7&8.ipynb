```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Load the image in grayscale
image = cv2.imread(r"C:\Users\rithi\Downloads\cone.jpg", cv2.IMREAD_GRAYSCALE)

# Step 2: Apply Canny edge detection
edges = cv2.Canny(image, 50, 150, apertureSize=3)

# Step 3: Apply Probabilistic Hough Line Transform to detect lines
lines = cv2.HoughLinesP(edges, 1, np.pi / 180, 50, minLineLength=100, maxLineGap=20)

# Step 4: Check if any lines are detected
if lines is not None:
    # Convert the grayscale image to BGR (color image) to draw colored lines
    color_image = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)

    for x1, y1, x2, y2 in lines[:, 0]:
        # Draw the line on the image (in red)
        cv2.line(color_image, (x1, y1), (x2, y2), (0, 0, 255), 5)

    # Step 5: Display the result
    plt.imshow(cv2.cvtColor(color_image, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for proper display in matplotlib
    plt.axis('off')  # Hide axis
    plt.show()

else:
    print("No lines were detected.")
```
